```python
# Unidad 5 Matrices

# Tablero de 3x3 (Matriz) representando el juego de TA-TE-TI

tablero = [
    ['X', ' ', 'O'],  # Fila 0
    ['O', 'X', ' '],  # Fila 1
    ['X', 'O', ' ']   # Fila 2
]

# Recorrido de la Matriz (Búsqueda de una 'X' en la diagonal principal)
print("Diagonal principal:")
for i in range(len(tablero)):
    # El valor en la diagonal principal es siempre (i, i)
    print(f"Posición ({i},{i}): {tablero[i][i]}") 

# El recorrido de la matriz con dos bucles 'for' anidados es una operación fundamental.
```

```python
# Unidad 6 Grafos y arboles

# Árbol de decisiones simple para recomendar bebidas

def recomendar_bebida(edad, hora):
    if edad < 18:
        if hora < 12:
            return "Jugo"
        else:
            return "Gaseosa"
    else:
        if hora < 18:
            return "Café"
        else:
            return "Vino"

print(recomendar_bebida(15, 19))


```

```python
# Unidad 6 Grafos

# Grafo: Representación de una red de amigos
# {'Persona': [Lista de sus amigos]}

red_amigos = {
    'Alicia': ['Bruno', 'Carlos', 'David'],
    'Bruno': ['Alicia', 'Evelin'],
    'Carlos': ['Alicia', 'Franco'],
    'David': ['Alicia', 'Evelin'],
    'Evelin': ['Bruno', 'David', 'Franco'],
    'Franco': ['Carlos', 'Evelin']
}

print("Nodos del Grafo (Personas):", list(red_amigos.keys())) # Cada persona es un nodo del grafo

print("Amigos de Alicia:", red_amigos['Alicia']) # lista de personas conectadas a Alicia
```
