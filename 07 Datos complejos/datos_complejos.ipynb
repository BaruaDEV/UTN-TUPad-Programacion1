```python
# Ejercicio 4

contactos = {'Ana': 1234567, 'Mora': 1236548}

def validar_nombre(nombre):
    """Retorna True si el nombre NO existe, False si SÍ existe."""
    if nombre in contactos:
        return False
    return True

def validar_numero(numero):
    """Retorna True si el número NO existe, False si SÍ existe."""
    if numero in contactos.values():
        return False
    return True

def almacenar_contacto():
    nombre = input("Ingrese nombre: ").strip().capitalize()
    
    # Valida que el nombre sea NUEVO
    if not validar_nombre(nombre):
        print("El contacto ya existe!")
        return
    
    # Manejo de error para la entrada del número
    while True:
        num_str= input("Ingrese numero: ")

        if num_str.isdigit():
            num=int(num_str)
            break
        else:
            print("Entrada invalida, ingresa solo digitos numericos")

    # Valida que el número sea NUEVO
    if not validar_numero(num):
        # Si validar_numero retorna False, el número ya existe
        print("El número ya está asociado a otro contacto!")
        return
    
    # Si todo es válido, almacena el contacto
    contactos[nombre] = num
    print(f"Contacto '{nombre}' almacenado exitosamente.")

def consultar_numero():
    nombre = input("Ingrese nombre de contacto: ").strip().capitalize()
    
    # Consulta directa de la clave en el diccionario
    if nombre in contactos:
        # Si la clave existe, imprime su valor
        print(f"El número de {nombre} es: {contactos[nombre]}")
    else:
        # Si la clave no existe
        print(f"El contacto '{nombre}' no existe!")

def menu():
    while True:
        print("\n::Eleji una opcion::\n")
        print("1. Almacenar contacto")
        print("2. Consultar numero")
        print("3. Salir")

        opcion = input("\nIngresa una opcion: ").capitalize()

        match opcion:
            case "1":
                almacenar_contacto()
            case "2":
                consultar_numero()
            case "3":
                print("\nFinalizando programa...")
                break
            case _:
                print("\nOpcion incorrecta, vuelva a intentarlo\n")


menu()
```

```python
# Ejercicio 5
def conteo_frase():
    frase = input("Ingresa una frase: ").lower()
    # Separamos todas las palabras por los espacios entre ellas
    palabras = frase.split()
    # set() para hacer ver las palabras unicas
    palabras_unicas = set(palabras)

    print(":: Palabras unicas")
    print(palabras_unicas)

    # Creamos el diccionario vacio
    cantidad_palabras = {}
    # iteramos sobre todas las palabras
    for i in palabras:
        # si la palabra se repite, a la cantidad se le suma 1
        if i in cantidad_palabras:
            cantidad_palabras[i] += 1
        # si aparece por primera vez la cantidad ser 1
        else:
            cantidad_palabras[i] = 1

    print(":: Cantidad de palabras")
    print(cantidad_palabras)

conteo_frase()
```

```python
# Ejercicio 6

def promedio_notas():

    alumnos = {}
    cant_alumnos = 3
    cant_notas = 3

    # Bucle para ingresar los 3 alumnos
    for i in range(cant_alumnos):
        while True:
            nombre = input(f"Ingresa nombre del alumno {i +1}: ").strip().capitalize()
            if nombre:
                break
            print("El nombre no puede estar vacio...")

        notas = []
        
        # bucle para ingresar las 3 notas
        for j in range(cant_notas):
            while True:
                nota_str=input(f"Ingrese nota numero {j +1}: ").strip()
                if not nota_str.isdigit():
                    print("Entrada invalida")
                    continue

                nota = int(nota_str)
                if 0 <= nota <= 10:
                    notas.append(nota)
                    break
                else:
                    print("La nota debe ser de 0 a 10")

    # Guardamos cada alumno con una tupla de sus notas
        alumnos[nombre]=tuple(notas)

    # Promedio de notas
    for nom, tupla_notas in alumnos.items():
        suma_notas=sum(tupla_notas)
        promedio= suma_notas / len(tupla_notas)
        print(f"El promedio de {nom} es de: {promedio:.2f}")

promedio_notas()
    

```

```python
# Ejercicio 7

# Definición de los sets de datos
aprobados_p1 = {101, 105, 110, 115, 120, 130}
aprobados_p2 = {105, 115, 125, 130, 135, 140}

print(f"Aprobaron Parcial 1: {aprobados_p1}")
print(f"Aprobaron Parcial 2: {aprobados_p2}")
print("-" * 40)

# 1. Aprobaron ambos parciales
ambos_parciales = aprobados_p1 & aprobados_p2
print(f"1. Aprobaron ambos parciales: {ambos_parciales}")

# 2. Aprobaron solo uno de los dos
solo_uno = aprobados_p1 ^ aprobados_p2
print(f"2. Aprobaron solo uno de los dos: {solo_uno}")

# 3. Lista total de estudiantes que aprobaron al menos un parcial
total_aprobados = aprobados_p1 | aprobados_p2
print(f"3. Total de estudiantes que aprobaron al menos uno: {total_aprobados}")
```

```python
# Ejercicio 8

stock_productos = {
    'Manzana': 50,
    'Leche': 20,
    'Pan': 35
}

def gestionar_stock():

    
    while True:
        print("\n::: Menú de Stock :::")
        print("1. Consultar stock de un producto")
        print("2. Agregar unidades / Agregar producto nuevo")
        print("3. Salir")
        
        opcion = input("\nSeleccione una opción: ").strip()

        match opcion:
            case '1':
                consultar_stock()
            case '2':
                modificar_stock()
            case '3':
                print("Finalizando programa...")
                break
            case _:
                print("Opción no válida. Intente nuevamente")

def consultar_stock():
    nombre_producto = input("Ingrese el nombre del producto a consultar: ").strip().capitalize()
    
    # Consulta directa de la clave en el diccionario
    if nombre_producto in stock_productos:
        stock_actual = stock_productos[nombre_producto]
        print(f"Stock de '{nombre_producto}': {stock_actual} unidades.")
    else:
        print(f"El producto '{nombre_producto}' no se encuentra en el inventario.")

def modificar_stock():
    nombre_producto = input("Ingrese el nombre del producto: ").strip().capitalize()
    
    # Bucle para validar que el usuario ingrese una cantidad válida
    while True:
        cantidad_str = input(f"Ingrese la cantidad a agregar al stock de {nombre_producto}: ").strip()
        
        # Validar que la entrada sea numérica y positiva
        if cantidad_str.isdigit():
            cantidad = int(cantidad_str)
            break
        else:
            print("Entrada inválida. Por favor, ingrese un número entero positivo.")
            

    if nombre_producto in stock_productos:
        # Agregar unidades al stock si el producto ya existe.
        stock_productos[nombre_producto] += cantidad
        print(f"Se agregaron {cantidad} unidades a '{nombre_producto}'")
    else:
        # Agregar un nuevo producto si no existe.
        stock_productos[nombre_producto] = cantidad
        print(f"Producto nuevo agregado: '{nombre_producto}' con {cantidad} unidades de stock.")

gestionar_stock()
```

```python
# Ejercicio 9

agenda = {
    ("lunes", "10:00"): "Reunion",
    ("martes", "15:00"): "Clase de ingles",
    ("miercoles", "14:00"): "Clase de programacion",
    ("jueves", "9:00"): "Gym",
    ("viernes", "19:00"): "Futbol"
}

def consultar_agenda():
    dia = input("Ingrese dia a consultar: ").strip().lower()
    hora= input("Ingrese hora a consultar (Ej: 10:00): ").strip().lower()

    clave_consulta = (dia , hora)

    if clave_consulta in agenda:
        evento = agenda[clave_consulta]
        print(f"Evento encotrado: {evento} el dia {dia} a las {hora} horas")
    else:
        print(f"No se encontro ningun evento el dia {dia} a la hora {hora} en la agenda")


consultar_agenda()
```

```python
# Ejercicio 10
paises_capitales = {
    "Argentina": "Buenos Aires", 
    "Chile": "Santiago"
}

def invertir_diccionario(diccionario):
    nuevo_diccionario = {}

    for pais, capital in diccionario.items():
        nuevo_diccionario [capital]=pais

    return nuevo_diccionario

diccionario_invertido = invertir_diccionario(paises_capitales)

print("Original")
print(paises_capitales)
print("*"*50)
print("*"*50)
print("Invertido")
print(diccionario_invertido)
```
