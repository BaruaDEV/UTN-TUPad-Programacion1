```python
# Ejercicio 4

productos = []

with open ("productos.txt", "r") as archivo:

    for i in archivo:
        linea_limpia = i.strip()
        datos= linea_limpia.split(',')
    
        if len(datos) == 3:
            nombre = datos[0]
            precio = float(datos[1])
            cantidad = int(datos[2])

            #creamos el diciconario con los datos
            dic_productos = {
                "nombre": nombre,
                "precio": precio,
                "cantidad": cantidad
            }
            # cargamos el diccionario creado a la lista productos
            productos.append(dic_productos)
        else:
            print(f"La linea no tiene 3 campos: {linea_limpia}")
```

```python
# Ejercicio 5

# una vez creado la lista en el punto 4, queda una lista de diccionario, como ejemplo:

productos = [
    {'nombre': 'Laptop', 'precio': 1200.5, 'cantidad': 5},
    {'nombre': 'Mouse', 'precio': 25.99, 'cantidad': 15},
    {'nombre': 'Monitor', 'precio': 350.0, 'cantidad': 8}
    # Asume que esta lista fue cargada previamente con la función anterior
]

# Podemos buscar los productos guardados por su nombre
buscar_producto = input("Ingrese nombre del producto: ").strip().lower()

encontrado = False
# Iteramos sobre la lista de diccionario y si es igual al nombre ingresado en la busqueda lo imprimimos
for p in productos:
    if p["nombre"].lower() == buscar_producto:
        print("\n::: Producto encontrado :::")
        print(f"Nombre: {p['nombre']}")
        print(f"Precio: {p['precio']}")
        print(f"Cantidad: {p['cantidad']}")

        encontrado = True
        break
# Sino imprimimos que no se ha encontrado el producto
if not encontrado:
    print(f"El producto {buscar_producto} no se encuentra disponible")
```

```python
# Ejercicio 6

# Una vez realizados el punto 3 para agregar productos, el punto 4 para guardarlas
# en una lista y el punto 5 para buscar el producto:
# Sobreescribimos el archivo:

# abrimos en modo escritura "w"
with open("productos.txt", "w") as archivo:
    # iteramos sobre todos los productos y guardamos en la variable 'linea'
    for p in productos:
        linea = f"{p['nombre'],{p['precio']},{p['cantidad']}}"
        # sobreescribimos todos los productos, inluyendo los agregados
        archivo.write(linea)

```
