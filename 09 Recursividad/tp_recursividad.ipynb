```python
# Ejercicio 1

def factorial(n):
    """
    Calcula el factorial de un número entero positivo 'n' de forma recursiva.
    """
    if n <= 1:
        return 1  
    else:
        return n * factorial(n - 1)

def mostrar_factoriales():
    """
    Función para solicitar el número al usuario y mostrar los factoriales.
    """
    max_num = int(input("Ingresa un número entero positivo: "))

    if max_num < 0:
        print("El número debe ser positivo.")
        return

    # Itera desde 1 hasta max_num para calcular e imprimir cada factorial
    for i in range(1, max_num + 1):
        resultado = factorial(i)
        print(f"Factorial de {i} ({i}!): {resultado}")


mostrar_factoriales()
```

```python
# Ejercicio 2

def fibonacci(n):
    """
    Calcula el valor de la serie de Fibonacci en la posición 'n' de forma recursiva.

    La serie comienza con F(0) = 0 y F(1) = 1.
    """
    # Caso Base 1: Posición 0
    if n <= 0:
        return 0
    
    # Caso Base 2: Posición 1
    elif n == 1:
        return 1
    
    # Paso Recursivo: F(n) = F(n-1) + F(n-2)
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

def mostrar_serie_fibonacci():
    """
    Solicita al usuario la posición máxima y muestra la serie completa.
    """
    max_pos = int(input("Ingresa la posición límite para la serie de Fibonacci: "))

    if max_pos < 0:
        print("La posición debe ser un número entero no negativo.")
        return

    serie = []
    
    # Iterar desde 0 hasta la posición máxima (inclusive)
    for i in range(max_pos + 1):
        valor = fibonacci(i)
        serie.append(str(valor))

    print(f"\nSerie de Fibonacci hasta la posición {max_pos}:")
    # Mostrar la serie como una cadena separada por comas
    print(", ".join(serie))


mostrar_serie_fibonacci()
```

```python
# Ejercicio 3

def potencia(base, exponente):
    """
    Calcula la potencia de un número base elevado a un exponente de forma recursiva.

    Argumentos:
        base (int/float): El número base (n).
        exponente (int): El exponente entero (m).
        
    Retorna:
        int/float: El resultado de la operación base^exponente.
    """
    # Caso Base 1: Exponente 0
    # Cualquier número elevado a 0 es 1 (n^0 = 1).
    if exponente == 0:
        return 1
    
    # Caso Base 2: Exponente 1
    # Cualquier número elevado a 1 es el mismo número (n^1 = n).

    if exponente == 1:
        return base
        
    # Manejo de Exponentes Negativos
    # n^(-m) = 1 / n^m
    if exponente < 0:
        # Llama a la función recursiva con el exponente positivo y luego calcula el inverso.
        return 1 / potencia(base, -exponente)

    # Paso Recursivo para Exponentes Positivos: n^m = n * n^(m-1)
    else:
        return base * potencia(base, exponente - 1)

def ejecutar_calculadora_potencia():
    """
    Algoritmo general para probar la función potencia con la entrada del usuario.
    """
    # Pide la base y el exponente al usuario (se asume entrada numérica válida)
    base = float(input("Ingresa la base (número a elevar): "))
    exponente = int(input("Ingresa el exponente (número entero): "))

    # Calcula el resultado
    resultado = potencia(base, exponente)

    # Muestra el resultado
    print(f"El resultado de {base} elevado a la potencia {exponente} es:")
    print(f"Resultado: {resultado}")


ejecutar_calculadora_potencia()
```

```python
# Ejercicio 4

def decimal_a_binario(n):
    """
    Convierte un número entero positivo en base decimal a su representación binaria 
    (cadena de texto) de forma recursiva.

    La conversión se basa en la división sucesiva por 2.

    Argumento:
        n (int): El número entero positivo en decimal.

    Retorna:
        str: La representación binaria de 'n'.
    """
    # Manejo de entrada no válida (se asume entero positivo, pero se añade un chequeo)
    if not isinstance(n, int) or n < 0:
        return "Error: Ingrese un número entero positivo."

    # Caso Base: Si n es 0, el binario es "0".
    if n == 0:
        return "0"
    
    # Caso Base: Si n es 1, el binario es "1".
    if n == 1:
        return "1"

    # Convertimos el resto a cadena para la concatenación
    resto = str(n % 2)
    
    # Llamada recursiva con el cociente
    cociente = n // 2
    
    return decimal_a_binario(cociente) + resto

def probar_conversion():
    """
    Función para solicitar el número al usuario y mostrar la conversión binaria.
    """
    print("--- Conversor Decimal a Binario Recursivo ---")
    
    # Pide el número al usuario
    try:
        numero_decimal = int(input("Ingresa un número entero positivo para convertir a binario: "))
    except ValueError:
        print("Entrada no válida. Por favor, ingresa un número entero.")
        return

    # Ejecuta la función recursiva
    resultado_binario = decimal_a_binario(numero_decimal)

    print(f"El número decimal: {numero_decimal}")
    print(f"La representación binaria es: {resultado_binario}")

probar_conversion()
```

```python
# Ejercicio 5

def es_palindromo(palabra):
    """
    Verifica si una cadena de texto es un palíndromo de forma recursiva.

    Un palíndromo se lee igual de adelante hacia atrás que de atrás hacia adelante.

    Argumento:
        palabra (str): La cadena de texto sin espacios ni tildes.

    Retorna:
        bool: True si es palíndromo, False en caso contrario.
    """
    # Caso Base 1: Una cadena con 0 o 1 carácter es siempre un palíndromo.
    # Esta es la condición de parada de la recursión.
    if len(palabra) <= 1:
        return True

    # Caso Base 2
    # Si el primer carácter es diferente al último, NO es un palíndromo.
    if palabra[0] != palabra[-1]:
        return False
    
    # Si los caracteres coinciden, se llama recursivamente a la función
    # con la subcadena que excluye el primer y el último carácter.
    else:
        return es_palindromo(palabra[1:-1])

def probar_palindromo():
    """
    Función para probar la función es_palindromo.
    """
    print("::: Detector de Palíndromos Recursivo :::")
    
    palabras = ["reconocer", "anilina", "neuquen", "casa", "oso", "holis"]
    
    print("Palabras de prueba")
    
    for palabra in palabras:
        resultado = es_palindromo(palabra)
        print(f"'{palabra}': {'Sí es un palíndromo' if resultado else 'No es un palíndromo'}")

# --- Ejecución del programa ---

probar_palindromo()
```

```python
# Ejercicio 6

def suma_digitos(n):
    """
    Calcula la suma de todos los dígitos de un número entero positivo 'n' 
    de forma recursiva, sin convertir el número a cadena de texto.

    La recursión se basa en dos operaciones clave:
    1. n % 10: Obtiene el último dígito (el resto).
    2. n // 10: Obtiene el resto del número (el cociente entero), que es la parte a sumar recursivamente.

    Argumento:
        n (int): El número entero positivo.

    Retorna:
        int: La suma de sus dígitos.
    """
    # Caso Base: Si el número tiene un solo dígito (o es 0), la suma es el número mismo.
    if n < 10:
        return n
    
    # Retorna el último dígito (n % 10) 
    # más la llamada recursiva con el resto del número (n // 10).
    else:
        # Último dígito (el que se suma)   +   Suma recursiva de los dígitos restantes
        return (n % 10) + suma_digitos(n // 10)

def probar_suma_digitos():
    """
    Función para probar la función suma_digitos con los ejemplos dados.
    """
    print("::: Suma de Dígitos Recursiva :::")
    
    ejemplos = [1234, 9, 305, 42]
    
    for num in ejemplos:
        resultado = suma_digitos(num)
        print(f"La suma de los dígitos de {num} es: {resultado}")


probar_suma_digitos()
```

```python
# Ejercicio 7

def contar_bloques(n):
    """
    Calcula el total de bloques necesarios para construir una pirámide de 'n' niveles 
    de forma recursiva.
    
    La suma es la de los primeros n números: n + (n-1) + ... + 1.

    Argumento:
        n (int): El número de bloques en el nivel más bajo (número de niveles). 
                 Debe ser un entero positivo.

    Retorna:
        int: El número total de bloques.
    """
    # Caso Base: Si n es 0 o 1, el número de bloques es n.
    # Usamos n=0 para detener la recursión, aunque la pirámide empieza en n=1.
    if n <= 0:
        return 0
    
    # La suma total es igual al número de bloques del nivel actual (n) 
    # más la suma de bloques de todos los niveles superiores (contar_bloques(n - 1)).
    else:
        return n + contar_bloques(n - 1)

def probar_contar_bloques():
    """
    Función para probar la función contar_bloques con los ejemplos dados.
    """
    print("::: Conteo de Bloques de Pirámide Recursivo :::")
    
    ejemplos = [1, 2, 4, 7]
    
    print("Pruebas de la función:")
    
    for num in ejemplos:
        resultado = contar_bloques(num)
        print(f"contar_bloques({num}) → {resultado}")

# Ejecución del programa

probar_contar_bloques()
```

```python
# Ejercicio 8

def contar_digito(numero, digito):
    """
    Cuenta cuántas veces aparece un dígito específico dentro de un número entero positivo 
    de forma recursiva.

    Argumentos:
        numero (int): El número entero positivo a analizar.
        digito (int): El dígito (entre 0 y 9) a contar.

    Retorna:
        int: La cantidad de veces que aparece el dígito.
    """
    # Caso Base: Si el número es 0, ya no hay más dígitos que contar.
    if numero == 0:
        return 0

    
    # 1. Obtener el último dígito del 'numero'.
    ultimo_digito = numero % 10
    
    # 2. Verificar si el último dígito coincide con el 'digito' buscado.
    # Si coincide, se suma 1. Si no, se suma 0.
    conteo_actual = 1 if ultimo_digito == digito else 0
    
    # 3. Llamada recursiva con el número sin el último dígito (numero // 10),
    # y se suma el conteo_actual.
    return conteo_actual + contar_digito(numero // 10, digito)

def probar_contar_digito():
    """
    Función para probar la función contar_digito con los ejemplos dados.
    """
    print("::: Contador de Dígitos Recursivo :::")
    
    ejemplos = [
        (12233421, 2, 3), 
        (5555, 5, 4),      
        (123456, 7, 0),
        (10101, 0, 2)
    ]
    
    for numero, digito, esperado in ejemplos:
        resultado = contar_digito(numero, digito)
        print(f"contar_digito({numero}, {digito}) → {resultado} (Esperado: {esperado})")



probar_contar_digito()
```
